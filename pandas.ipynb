Series

```python
num_list = [1,2,3,4,5]
pandas_Series1 = pd.Series(num_list)
print(pandas_Series1)
```

Series using coustom index

```python
num_list = [1,2,3,4,5]
index = ['a','b','c','d','e']
pSeries2 = pd.Series(num_list,index)
print(pSeries2),type(pSeries2)
```
