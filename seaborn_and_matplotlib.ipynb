Matplotlib is a Python library for creating charts and graphs.

Importig

```python
import matplotlib.pyplot as plt

# This tells Jupyter to show plots in the notebook
%matplotlib inline
```

Line graph


*   common colors: 'red', 'blue', 'green', 'purple', 'orange'


*   Common markers: 'o' (circle), 's' (square), '^' (triangle), '*' (star)

*   Line styles: '-' (solid), '--' (dashed), '-.' (dash-dot), ':' (dotted)





```python
# Data: days and hours studied
days = [1, 2, 3, 4, 5]
hours = [2, 3, 4, 3, 5]

# Create the plot
plt.plot(days, hours,color='red', linestyle=':',marker='*')

#making graph professional
plt.xlabel('Days')
plt.ylabel('Hours')
plt.title('Hours studied per day')

#Adding graph lines
plt.grid(True)

# Display the plot
plt.show()
```

**Bar graph**

*   plt.bar (vertical)


*   plt.barh(horizontal







```python
fruits=['apple','banana','cherry']
quantity=[50,60,20]
plt.bar(fruits,quantity, color='green', edgecolor='black')
plt.title('fruits and sales')
plt.xlabel('fruits')
plt.ylabel('sales')
plt.show()
```

```python
vegetables=['cabbage','carrot','spinach']
quantity=[50,60,20]
plt.barh(vegetables,quantity, color='pink')
plt.title('vegetables and sales')
plt.xlabel('sales')
plt.ylabel('vegetables')
plt.title( 'vegetable sales this week')
plt.show()
```

**Scatter plots**

```python
calories_per_day=[2000,2500,3000,3500]
steps_per_day=[6000,7000,8000,9000]
plt.scatter(calories_per_day,steps_per_day,color='green',marker='o',s=90)
plt.xlabel('calories')
plt.ylabel('steps')
plt.title('steps vs calories per day')
plt.grid(True)
plt.show()
```

Histograms

*   bins(How many divisions)


*   for height, student scores





---



```python
std_scores=[90,89,90,67,78,56,78,90,45,76,87,98,84,85,86,94,56,87,65,98,78,40,50,60]
plt.hist(std_scores,bins=8,color='blue',edgecolor='black')
plt.xlabel=('student scores')
plt.ylabel=('frequency')
plt.title('histogram of student scores')
plt.show()
```

pie charts

```python
activities = ['Sleep', 'Work', 'Exercise', 'Leisure']
hours = [5, 8, 2, 4]
explode=(0,0.1,0,0)
plt.pie(hours, labels=activities, autopct='%1.1f%%', explode=explode)
plt.title('Daily Time Distribution')

# plt.legend()
plt.show()
```

**Seaborn** is a Python visualization library based on Matplotlib that provides a high-level interface for drawing attractive statistical graphics.

Importing

```python
import seaborn as sns
import matplotlib.pyplot as plt
import pandas as pd
import numpy as np

%matplotlib inline

print("Seaborn version:", sns.__version__)
```

Built id datasets

```python
titanic = sns.load_dataset('titanic')

# Show first few rows
print(titanic.head())
print("\nDataset shape:", titanic.shape)
print("\nColumn info:")
print(titanic.info())
```

```python
sns.scatterplot(data=titanic, x='survived', y='pclass')
plt.title('passenger survived vs passsenger class')
plt.show()
```

```python
# Load the famous tips dataset
tips = sns.load_dataset('tips')

# Show first few rows
print(tips.head())
print("\nDataset shape:", tips.shape)
print("\nColumn info:")
print(tips.info())
```

```python
sns.scatterplot(data=tips, x='total_bill', y='tip')
plt.title('Bill Amount vs Tip')
plt.show()
```

```python
sns.scatterplot(data=tips, x='total_bill', y='tip', hue='time')
plt.title('Bill vs Tip (colored by time)')
plt.show()
```

countplot

```python
sns.histplot(data=tips, x='total_bill', bins=20)
plt.title('Distribution of Bill Amounts')
plt.show()
```

```python
sns.pairplot(tips, hue='time')
plt.show()
```
